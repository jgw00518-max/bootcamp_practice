```python
import sqlite3 # sqlite는 관리용 (퍼포먼스 빠름), 서비스용은 아님

# sqlite 데이터베이스 연결하기
dbPath = "../Data/test.sqlite"
conn = sqlite3.connect(dbPath) # connect는 연결 역할

# 테이블을 생성하고 데이터 넣기
curs = conn.cursor() # 데이터베이스 작업을 하려면 cursor 필요
curs.executescript(  # excute = 한줄, executescript = 여러줄
    """ 
    DROP TABLE IF EXISTS items;
    CREATE TABLE items(
        item_id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        price INTEGER
    );
    """ # 줄이 끝나는 위치에 ;
)

# 위의 조작을 데이터베이스에 반영하기
conn.commit()
```

```python
# 데이터 넣기
curs = conn.cursor()
curs.execute(
    "INSERT INTO items (name, price) VALUES (?,?)", # 테이블의 한 줄을 레코드 혹은 튜플이라고 한다.
    ("Orange", 5200)                                # 따라서 데이터를 튜플로 넣어준다.
)
conn.commit()
```

```python
# 여러 데이터 연속으로 넣기
curs = conn.cursor()
data = [("Mango", 7700), ("Kiwi", 4000), ("Grape", 8700), ("Peach", 9400)]
curs.executemany(   # 반복문 역할
    "INSERT INTO items (name, price) VALUES (?,?)",
    data
)
conn.commit()
```

```python
# 데이터 추출하기
curs = conn.cursor()
curs.execute(
    "SELECT item_id, name, price FROM items"
)
item_list = curs.fetchall() # 리스트 가져오기
item_list
# for item in item_list:
#     print(item)
```

```python
# 4000원 ~ 7000원 사이의 데이터 추출하기
curs = conn.cursor()
curs.execute(
    "SELECT * FROM items WHERE price BETWEEN ? AND ?",
    (4000, 7000)    # [] , () 상관 없음 - 데이터베이스는 구분 못함
)
item_list = curs.fetchall()
item_list
```

```python
from tinydb import TinyDB, Query, where

# 데이터베이스 연결하기
filepath = "../Data/tinydb.json" # 문서형 데이터베이스는 모두 json 형식
db = TinyDB(filepath)
```

```python
# 기존에 테이블이 있다면 제거하기
db.drop_table('fruits')
```

```python
# 테이블 생성하기
table = db.table('fruits')
```

```python
# 테이블에 데이터 추가하기
table.insert(
    {
        'name' : 'Banana',
        'price' : 6000
    }
)
table.insert(
    {
        'name' : 'Orange',
        'price' : 12000
    }
)
table.insert(
    {
        'name' : 'Mango',
        'price' : 7000
    }
)
```

```python
# 모든 데이터 추출하기
print(table.all())
```

```python
# Orange 검색하기
item = Query()
res = table.search(item.name == 'Orange')
print(res)
print(f"orange is {res[0]['price']}.")
```

```python
# 가격이 8000이상인 것 추출하기
item = Query()
res = table.search(item.price >= 8000)
res
```

```python
# 수정하기
table.upsert({'price':8000}, item.name == "Mango")
res = table.search(item.name == "Mango")
res
```
