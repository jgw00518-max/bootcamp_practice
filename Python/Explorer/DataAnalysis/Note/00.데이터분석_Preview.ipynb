### Pandas를 이용하여 데이터프레임 만들기

```python
# 모듈 불러오기
import pandas as pd
```

```python
# List로 DataFrame 만들기
temp = \
[
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

df = pd.DataFrame(temp)
df
```

### DataFrame에서 값 추출하기

```python
# 열 기준으로 추출하기
df[0]
df.loc[:,0]
```

```python
# 행 기분으로 추출하기
df.loc[0,:]
```

```python
# 범위를 정해 열 기준으로 추출하기
df.loc[:,0:1] # Slicing
df.loc[:,[0,1]] # Indexing
df.loc[:,range(0,1+1)] # range 함수
```

```python
# 범위를 정해 행 기준으로 추출하기
df.loc[:1,:] # 0 생략 가능
df.loc[[0,1],:]
df.loc[range(0,1+1),:]
```

```python
# 특정 위치의 값만 추출하기
print(df.loc[1,1])
```

```python
temp = \
{
    'weight' : [80.0, 70.4, 65.5, 45.9, 51.2],
    'height' : [ 170,  180,  155,  143,  154],
    'type'   : [ 'f',  'n',  'n',  't',  't']
}
tbl = pd.DataFrame(temp)
tbl
```

### 반복을 통한 집계

```python
# 계산하기 - 1 : 1880년의 여자아이의 합계를 구하기

names[(names.year == 1880) & (names.gender == 'F')].births.sum()
```

```python
# 년도의 범위
sorted(names.year.unique()) # 중복되지 않은 값만 뽑아오기 (unique)
```

```python
# 계산하기 - 2 : 여자아이의 년도별 합계를 구하기
female = []
for year in sorted(names.year.unique()):
    count = 0
    count = int(names[(names.year == year) & (names.gender == 'F')].births.sum())
    female.append(count)

print(female)
```

```python
# 남자아이와 여자아이를 적용
female = []
male = []

for year in sorted(names.year.unique()):
    count = 0
    count = int(names[(names.year == year) & (names.gender == 'F')].births.sum())
    female.append(count)

    count = int(names[(names.year == year) & (names.gender == 'M')].births.sum())
    male.append(count)

print(female)
print(male)
```

```python
# DataFrame 만들기
total_births = pd.DataFrame(
    {
        'year' : sorted(names.year.unique()),
        'F' : female,
        'M' : male
    }
)
total_births.set_index('year', inplace=True)
total_births.head()
```

```python
# 선그래프 생성하기
total_births.plot()
plt.show()
```

```python
# 그래프의 가로 세로 비율 적용
plt.rcParams['figure.figsize'] = (14, 6)
```

```python
total_births.plot(
    title='Total births by gender and year'
)
plt.show()
```

#### 순수코딩

```python
# 연도와 성별로 정렬
names2 = names.sort_values(by=['gender', 'year'])
names2.head()
```

```python
female = []
male = []

for gender, births, year in zip(names2.gender, names2.births, names2.year):
    if gender == 'F':
        try:
            female[year-1880] += births
        except:
            female.append(births)
    else:
        try:
            male[year-1880] += births
        except:
            male.append(births)

print(female)
print(male)
```
