### DataFrame merging(병합)

```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {
        'key' : list('bbacaab'),
        'data1' : range(7)
    }
)
df1
```

```python
df2 = pd.DataFrame(
    {
        'key' : list('abd'),
        'data1' : range(3)
    }
)
df2
```

```python
# 병합하기
pd.merge(df1, df2, on='key', how='inner')
```

```python
# 전체 보기

pd.merge(df1, df2, on='key', how='outer')
```

```python
pd.merge(df1, df2, on='key', how='left')
```

```python
pd.merge(df1, df2, on='key', how='right')
```

```python
df3 = pd.DataFrame(
    {
        'lkey' : list('bbacaab'),
        'data1' : range(7)
    }
)
df3
```

```python
df4 = pd.DataFrame(
    {
        'rkey' : list('abd'),
        'data2' : range(3)
    }
)
df4
```

```python
pd.merge(df3, df4, left_on='lkey', right_on='rkey', how='inner')
```

```python
import numpy as np
```

```python
df1 = pd.DataFrame(
    np.arange(6).reshape(3,2),
    index=['a','b','c'],
    columns=['one', 'two']
)
df1
```

```python
df2 = pd.DataFrame(
    5+np.arange(4).reshape(2,2),
    index=['a','c'],
    columns=['three', 'four']
)
df2
```

```python
pd.concat([df1,df2], axis='columns', sort=True)
```

#### 데이터의 범위(구간) 정하기

```python
ages=[20,22,25,27,21,23,37,31,61,45,41,32]
bins=[18,25,35,60,100]
```

```python
cats = pd.cut(ages, bins)
cats
```

```python
cats.value_counts()
```

```python
group_names = ['Youth', 'YoungAdult', 'MiddleAged', 'Senior']
cats2 = pd.cut(ages, bins, labels=group_names)
cats2.value_counts()
```
