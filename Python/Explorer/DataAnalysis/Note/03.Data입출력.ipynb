#### 데이터 저장하기

```python
data = {
        'name' : [ 'Jerry', 'Riah', 'Paul'],
        'algol' : [ "A", "A+", "B"],
        'basic' : [ "C", "B", "B+"],
        'c++' : [ "B+", "C", "C+"],
        }

data
```

```python
df = pd.DataFrame(data).set_index('name')
df
```

```python
# csv로 저장하기
df.to_csv("../Data/df_sample.csv")
```

```python
pd.read_csv("../Data/df_sample.csv")
```

```python
# json으로 저장하기
df.to_json("../Data/df_sample.json")
```

```python
# Excel로 저장하기
df.to_excel("../Data/df_sample.xlsx")
```

### 여러 개의 데이터프레임을 하나의 Excel File로 저장

```python
writer = pd.ExcelWriter("../Data/df_excelwriter.xlsx")
df.to_excel(writer, sheet_name='sheet1')
df1.to_excel(writer, sheet_name='발전전력량')
writer.close()
```
