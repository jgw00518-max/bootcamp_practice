```python
# 고령자 정렬
pop_Seoul.sort_values(
    by='고령자',
    ascending=False
).head()
```

(정리)    
인구수가 제일 많은 지역은 송파구이지만, 영등포구가 제일 외국인이 많다.    
그리고 그 근처라고 할 수 있는 구로구, 금천구 등도 외국인이 많다.    
송파구는 고령자가 제일 많지만, 고령자의 비율은 높지 않은 것으로 확인된다.    
하지만 강북구, 도봉구, 중구, 종로구, 은평구 같은 경우는 인구수 대비    
고령자 비율이 높기 때문에 추후 고령화가 우려된다.

#### 인구대비 CCTV 비율

```python
plt.style.use('ggplot')

# 인구 1만 명당 CCTV 수
data_result['인구 1만명당 CCTV'] = (
    data_result['소계'] / data_result['인구수'] * 10000
)

# 오름차순 정렬
cctv_sorted = data_result['인구 1만명당 CCTV'].sort_values()

# 중구만 다른 색상 지정
colors = [
    'salmon' if gu == '중구' else 'skyblue'
    for gu in cctv_sorted.index
]

# 가로 막대그래프
ax = cctv_sorted.plot(
    kind='barh',
    figsize=(10, 10),
    color=colors
)

# 막대 끝에 소수점 둘째 자리까지 표시
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.2f', padding=3)

plt.title('서울시 구별 인구 1만 명당 CCTV 수')
plt.xlabel('인구 1만 명당 CCTV 수')
plt.ylabel('구별')

# 가장 긴 막대의 숫자가 잘리지 않도록 여유 공간 추가
plt.xlim(0, data_result['인구 1만명당 CCTV'].max() * 1.15)

plt.tight_layout()
plt.show()
```

#### 머신러닝의 알고리즘 중 하나인 선형회귀식을 이용한 인구수와 소계에 따른 회귀선 만들기
- 인구수 대비 가장 적정한 CCTV 대수를 학습한다.
- 선형회귀는 직선(Y=aX+b)이나 곡선의 함수식을 이용하여 구하는 방식

```python
from sklearn.linear_model import LinearRegression # anaconda는 기본으로 깔려있다.
```

```python
# 객체 생성
lr = LinearRegression()
```

```python
# 학습하기
lr.fit(
    data_result[['인구수']],
    data_result[['소계']]
)
```

```python
# 기울기와 절편
print(lr.coef_, lr.intercept_)
```

```python
# 50개의 X축 데이터 정하기
data_x = np.linspace(100000, 700000, num=50)
pred_x = pd.DataFrame({'인구수' : data_x})
pred_x.head()
```

```python
plt.style.use('ggplot')
plt.figure(figsize=(14,10))
plt.scatter(
    data_result['인구수'],
    data_result['소계']
)

# 회귀선 그리기
plt.plot(
    data_x,
    lr.predict(pred_x), # 예측을 위해선 DataFrame으로 데이터를 줘야함
    ls='dashed',
    lw=3,
    color='g'
)

plt.title('인구수와 CCTV설치 대수의 산점도')
plt.xlabel('인구수')
plt.ylabel('CCTV 대수')
plt.xlim(0,800000)
plt.ylim(0,6000)
plt.show()
```

```python
# Predict 연습
aaa = pd.DataFrame({'인구수' : [100000]})
lr.predict(aaa)
```

(결론)
- 이 직선은 CCTV대수의 기준값이 된다.
- 이 직선을 기준으로 상위에 있는 점은 CCTV가 과하게 설치된 것이고, 직선보다 하위에 있는 점은 인구수에 비해 CCTV가 부족하다는 해석이다.

#### 직선과 오차를 구하는 공식과 오차가 큰 순으로 데이터 정렬 후 그래프 그리기

```python
# 자치구별 선형회귀식의 예측값
print(lr.predict(data_result[['인구수']]))
```

```python
# 소계에서 기준 인구수 차이로 오차값 구하기
data_result['오차'] = np.abs(data_result['소계'] - lr.predict(data_result[['인구수']]).ravel())

df_sort = data_result.sort_values(by='오차', ascending=False)
df_sort.head(20)
```

```python
# 시각화 하기 : 오차가 큰 10개 구에 텍스트 붙이기
plt.style.use('ggplot')
plt.figure(figsize=(14,10))

scatter = \
plt.scatter(
    data_result['인구수'],
    data_result['소계'],
    c=data_result['오차']
)

# 회귀선 그리기
plt.plot(
    data_x,
    lr.predict(pred_x), # 예측을 위해선 DataFrame으로 데이터를 줘야함
    ls='dashed',
    lw=3,
    color='g'
)

# 강남구와 도봉구 점 아래에 CCTV 개수 표시
for gu in ['강남구', '도봉구']:
    plt.annotate(
        f"{data_result.loc[gu, '소계']:,.0f}",
        xy=(
            data_result.loc[gu, '인구수'],
            data_result.loc[gu, '소계']
        ),
        xytext=(0, -10),
        textcoords='offset points',
        ha='center',
        va='top',
        fontsize=12,
    )

for n in range(10):
    plt.annotate(
        df_sort.index[n],
        xy=(df_sort['인구수'].iloc[n], df_sort['소계'].iloc[n]+100),
        rotation=0,
        va='baseline',
        ha='center',
        fontsize=12
    )

plt.title('인구수와 CCTV설치 대수의 산점도')
plt.xlabel('인구수')
plt.ylabel('CCTV 대수')
plt.xlim(0,800000)
plt.ylim(0,6000)
plt.colorbar(scatter)
plt.show()
```
