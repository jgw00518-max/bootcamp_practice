### 행과 열 이름 변경하기

```python
# index와 column 이름 변경하기 - 1
df.index = ['학생1', '학생2']
df.columns = ['연령', '남녀', '소속']
df
```

```python
df = pd.DataFrame(
    [
        [15, '남', '덕영중'],
        [17, '여', '수리중']
    ],
    index=['준서','예은'],
    columns=['나이', '성별', '학교']
)
df
```

```python
# 나이 -> 연령
df.rename(
    columns={
        '나이': '연령'
    },
    inplace=True
)
df
```

```python
# 준서 -> 학생1
df.rename(
    index={
        '준서' : '학생1'
    },
    inplace=True
)
df
```

#### 행 삭제

```python
exam_data = {
    '수학' : [90, 80, 70],
    '영어' : [98, 89, 95],
    '음악' : [85, 95, 100],
    '체육' : [100, 90, 90]
}
exam_data
```

```python
df = pd.DataFrame(
    exam_data,
    index=['서준', '우현', '인아']
)

df
```

```python
# df의 복사
df2 = df.copy()
df2
```

```python
# df2에서 index가 우현인 행을 삭제
df2.drop(
    '우현',
    axis='rows',
    inplace=True
)
df2
```

```python
df2.shape # 중요. 나중에 차원을 이용할 때 쓰인다.
```

```python
# df2에서 index가 우현인 행을 삭제
df2.drop(
    '우현',
    axis=0, # 옛날 방식
    inplace=True
)
df2
```

```python
df3 = df.copy()
df3
```

```python
# 우현과 인아의 행 삭제하기
df3.drop(
    ['우현','인아'],
    axis='rows',
    inplace=True
)
df3
```

#### 열 삭제

```python
df4 = df.copy()
df4
```

```python
# 수학열을 삭제하기
df4.drop(
    '수학',
    axis='columns',
    inplace=True
)
df4
```

```python
df5 = df.copy()
df5
```

```python
# 영어 음악 컬럼 삭제하기
df5.drop(
    ['영어', '음악'],
    axis='columns',
    inplace=True
)
df5
```

#### 행 선택
- 행의 이름을 사용하는 loc
- 행의 위치를 사용하는 iloc

```python
# index가 서준인 행을

# loc 사용
df.loc['서준']

# iloc 사용
df.iloc[0]
```

```python
# 행 인덱스를 사용하여 서준과 우현 2개의 행을 추출
# loc
df.loc[['서준','우현']]
# iloc
df.iloc[[0,1]]
#-------------------------------------
# 행 인덱스 범위를 사용하여 서준부터 우현까지의 행 추출
# loc
df.loc['서준':'우현']
# iloc
df.iloc[:1+1]
```
