### Matplotlib
: 파이썬의 표준 시각화 도구

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 시도별 전출입 인구수 데이터를 df로 불러오기
df = pd.read_excel("../Data/시도별 전출입 인구수.xlsx")
df.head()
```

```python
df.tail()
```

```python
df.shape
```

```python
df.info()
```

```python
# 0번 Index 삭제하기
df.drop(
    0,
    axis='rows',
    inplace=True
)

df.head()
```

```python
# Index 재정리
df.reset_index(
    drop=True,
    inplace=True
)
df.head()
```

```python
# NaN 정리
df['전출지별'] = df['전출지별'].ffill() # 맨앞에 있는 column으로 채운다. / bfill() -> 맨뒤에 있는 column으로 채우기 (NaN 기준)
df.head()
```

```python
# Column의 Type
df.columns
```

```python
# 1970년 데이터의 내용 보기
df['1970'].unique()
```

```python
# - -> 0으로 변경
df.replace(
    '-',
    0,
    inplace=True
)
df.head()
```

```python
df.info()
```

----
#### 전출입 시각화

```python
# 서울에서 타지역으로 이동한 데이터를 추출
df_seoul = \
    df[(df.전출지별 == '서울특별시') & (df.전입지별 != '서울특별시')]
df_seoul.head()
```

```python
# 전출지별 컬럼 삭제
df_seoul.drop(
    '전출지별',
    axis='columns',
    inplace=True
)
df_seoul.head()
```

```python
# 전입지별 -> 전입지 변경
df_seoul.rename(
    {
        '전입지별' : '전입지'
    },
    axis='columns',
    inplace=True
)
df_seoul.head()
```

```python
# 전입지를 index로 변경
df_seoul.set_index(
    '전입지',
    inplace=True
)
df_seoul.head()
```

####  서울에서 경기도로 이동한 인구의 선그래프

```python
# 서울에서 경기도로 이동한 데이터 추출
sr_one = df_seoul.loc['경기도']
sr_one
```

```python
# 한글화 matplotlib 설치
# %pip install koreanize-matplotlib

import koreanize_matplotlib
```

```python
# 그래프 사이즈 조정(가로 : 14인치, 세로 : 5인치)
plt.figure(
    figsize = (14,5)
)

# 선그래프
plt.plot(
    sr_one
)

# 차트 제목
plt.title("서울 -> 경기 이동")

# 축 이름
plt.xlabel('기간')
plt.ylabel('이동 인구수')

# x축 눈금 라벨 회전하기
plt.xticks(
    rotation = 45
)

# y축 눈금 라벨 회전하기
plt.yticks(
    rotation = -15
)

# 범례
plt.legend(
    labels = ['서울 -> 경기'],
    loc = 'best' # best, upper left/right, lower left/right
)

plt.show()
```

```python
# matplotlib의 Style 종류
plt.style.available
```

```python
# 스타일 서식 지정
plt.style.use('ggplot')

# 그래프 사이즈 조정(가로 : 14인치, 세로 : 5인치)
plt.figure(
    figsize = (14,7)
)

# 선그래프
plt.plot(
    sr_one,
    zorder = 1
)

# 특정 marker 표시
plt.scatter(
    sr_one.index,
    sr_one.values,
    s= 100,
    c=['red' if i == 1995 - 1970 else 'blue' for i in range(2017-1970+1)]
)

# 차트 제목
plt.title("서울 -> 경기 이동", size = 20)

# 축 이름
plt.xlabel('기간', size=20)
plt.ylabel('이동 인구수', size=20)

# x축 눈금 라벨 회전 및 크기 조절
plt.xticks(
    rotation = 45,
    size = 7
)

# y축 눈금 라벨 회전 및 크기 조절
plt.yticks(
    rotation = -15,
    size = 10
)

# 범례
plt.legend(
    labels = ['서울 -> 경기'],
    loc = 'best', # best, upper left/right, lower left/right
    fontsize=20
)

# y축 범위 지정
plt.ylim(0, 800000)

# 주석 표시 - 화살표
plt.annotate(
    "",
    xy=(1990-1970, 620000), # 화살표의 머리 부분
    xytext=(2, 290000),     # 화살표의 꼬리 부분
    xycoords='data',        # 좌표 체계
    arrowprops=dict(arrowstyle="->", color='skyblue', lw=5), # 화살표 서식
    va = 'baseline', # 텍스트 상하 정렬
    ha = 'center'    # 텍스트 좌우 정렬
)

# 주석 표시 - 텍스트
plt.annotate(
    '인구 이동 증가(1970~1995)', # 텍스트 입력
    xy=(1980-1970, 400000),    # 텍스트 위치 기준점
    rotation = 31,             # 텍스트 회전 각도
    va = 'baseline',
    ha = 'center',
    fontsize=12
)

plt.annotate(
    "",
    xy=(2017 - 1970, 450000),      # 화살표 머리: 2017년
    xytext=(2000 - 1970, 590000),  # 화살표 꼬리: 1995년
    xycoords='data',
    arrowprops=dict(arrowstyle="->", color='olive', lw=5),
    va = 'baseline',
    ha = 'center'
)

plt.annotate(
    '인구 이동 감소(1995~2017)', 
    xy=(2008-1970, 520000),    
    rotation = -14,             
    va = 'baseline',
    ha = 'center',
    fontsize=12
)

plt.annotate(
    "",
    xy=(0, 130000),
    xytext=(0, 220000),
        xycoords='data',
    arrowprops=dict(arrowstyle="->", color='red', lw=4),
    va = 'baseline',
    ha = 'center'
)

# 1995년 데이터 표시
plt.annotate(
    f"{sr_one['1995']:,}명",
    xy=(1995-1970, sr_one['1995']+30000),
    rotation=0,
    va='baseline',
    ha='center',
    fontsize=12
)

plt.annotate(
    f"{sr_one['2017']:,}명",
    xy=(2017-1970, sr_one['2017']-35000),
    rotation=0,
    va='baseline',
    ha='center',
    fontsize=12
)

plt.annotate(
    f"{sr_one['1970']:,}명",
    xy=(0, sr_one['1970']+100000),
    rotation=0,
    va='baseline',
    ha='center',
    fontsize=12
)
plt.show()
```

```python
# 서울에서 충청남도, 경상북도, 강원도로 이동한 인구 데이터 값을 그래프
df_3 = \
df_seoul.loc[['충청남도', '경상북도', '강원도'],:]
df_3
```

```python
plt.style.use('ggplot')

fig = plt.figure(figsize=(20,5))
ax = fig.add_subplot(1,1,1)

ax.plot(
    df_3.loc['충청남도',:],
    marker='o',
    markerfacecolor='green',
    markersize=10,
    color='olive',
    linewidth=2,
    label='서울->충남'
)

ax.plot(
    df_3.loc['경상북도', :],
    marker='o', markerfacecolor='blue', markersize=10,
    color='skyblue', linewidth=2, label='서울->경북'
)

ax.plot(
    df_3.loc['강원도', :],
    marker='o', markerfacecolor='red', markersize=10,
    color='hotpink', linewidth=2, label='서울->강원'
)

# 제목과 축 설정
ax.set_title('서울 -> 충남, 경북, 강원 인구 이동')
ax.set_ylabel('이동 인구수')
ax.set_xlabel('기간')
ax.tick_params(axis='x', labelrotation=90)
ax.set_ylim(0, 70000)

# 세 지역 중 최대값에 빨간 + 표시
values = df_3.apply(pd.to_numeric)
region, year = values.stack().idxmax()
ax.scatter(
    year, values.loc[region, year],
    marker='+', color='red', s=400,
    linewidths=3, zorder=1
)

ax.legend(loc="best")

plt.show()
```

```python
plt.style.use('ggplot')

fig, axes = plt.subplots(2, 2, figsize=(20, 10))
fig.suptitle('서울 -> 충남,경북,강원,전남 인구 이동', fontsize=18)
df_j = df_seoul.loc['전라남도']
values = df_3.apply(pd.to_numeric)
values = pd.concat([values, pd.to_numeric(df_j).rename(df_j.name).to_frame().T])

# 서울 -> 충남
ax = axes[0, 0]
ax.plot(df_3.loc['충청남도'], marker='o', markerfacecolor='green',
        markersize=6, color='olive', linewidth=2, label='서울->충남')
ax.set_title('서울 -> 충남 인구 이동')
ax.set_xlabel('기간')
ax.set_ylabel('이동 인구수')
ax.set_ylim(0, 70000)
ax.tick_params(axis='x', labelrotation=90)
ax.legend(loc='best')
max_year_1 = values.loc['충청남도'].idxmax()
ax.scatter(max_year_1, values.loc['충청남도', max_year_1],
        marker='+', color='red', s=250, linewidths=3, zorder=1)
ax.annotate(
        f"{values.loc['충청남도', max_year_1]:,}명",
        xy=(max_year_1, values.loc['충청남도', max_year_1]),
        xytext=(0, 10), textcoords='offset points',
        ha='center', va='bottom', fontsize=10
)

# 서울 -> 경북
ax = axes[0, 1]
ax.plot(df_3.loc['경상북도'], marker='o', markerfacecolor='blue',
        markersize=6, color='skyblue', linewidth=2, label='서울->경북')
ax.set_title('서울 -> 경북 인구 이동')
ax.set_xlabel('기간')
ax.set_ylabel('이동 인구수')
ax.set_ylim(0, 70000)
ax.tick_params(axis='x', labelrotation=90)
ax.legend(loc='best')
max_year_2 = values.loc['경상북도'].idxmax()
ax.scatter(max_year_2, values.loc['경상북도', max_year_2],
        marker='+', color='red', s=250, linewidths=3, zorder=1)
ax.annotate(
        f"{values.loc['경상북도', max_year_2]:,}명",
        xy=(max_year_2, values.loc['경상북도', max_year_2]),
        xytext=(0, 10), textcoords='offset points',
        ha='center', va='bottom', fontsize=10
)

# 서울 -> 강원
ax = axes[1, 0]
ax.plot(df_3.loc['강원도'], marker='o', markerfacecolor='red',
        markersize=6, color='red', linewidth=2, label='서울->강원')
ax.set_title('서울 -> 강원 인구 이동')
ax.set_xlabel('기간')
ax.set_ylabel('이동 인구수')
ax.set_ylim(0, 70000)
ax.tick_params(axis='x', labelrotation=90)
ax.legend(loc='best')
max_year_3 = values.loc['강원도'].idxmax()
ax.scatter(max_year_3, values.loc['강원도', max_year_3],
        marker='+', color='hotpink', s=250, linewidths=3, zorder=1)
ax.annotate(
        f"{values.loc['강원도', max_year_3]:,}명",
        xy=(max_year_3, values.loc['강원도', max_year_3]),
        xytext=(0, 10), textcoords='offset points',
        ha='center', va='bottom', fontsize=10
)

# 서울 -> 전남
ax = axes[1, 1]
ax.plot(df_j, marker='o', markerfacecolor='orange',
        markersize=6, color='yellow', linewidth=2, label='서울->전남')
ax.set_title('서울 -> 전남 인구 이동')
ax.set_xlabel('기간')
ax.set_ylabel('이동 인구수')
ax.set_ylim(0, 70000)
ax.tick_params(axis='x', labelrotation=90)
ax.legend(loc='best')
max_year_j = values.loc['전라남도'].idxmax()
ax.scatter(max_year_j, values.loc['전라남도', max_year_j],
        marker='+', color='red', s=250, linewidths=3, zorder=1)
ax.annotate(
        f"{values.loc['전라남도', max_year_j]:,}명",
        xy=(max_year_j, values.loc['전라남도', max_year_j]),
        xytext=(0, 10), textcoords='offset points',
        ha='center', va='bottom', fontsize=10
)

plt.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

#### 산점도 (산포도, 스캐터)
: 서로 다른 두 단위 데이터의 관계를 나타낸다.

```python
# 연비(mpg)와 차중(weight)에 대한 상관계수 확인하기
df[['mpg', 'weight']].corr()
```

```python
# 연비(mpg)와 차중(weight)에 대한 산점도 그리기
plt.style.use('ggplot')

df.plot(
    kind='scatter',
    x='weight',
    y='mpg',
    figsize=(10,5),
    color=df.duplicated(subset=['weight', 'mpg'], keep=False).map(
        {True: 'blue', False: 'coral'}
    ),
    alpha=0.5,  # 투명도 조절 (중복 데이터를 구분하기 위함)
    s=50        # 점 크기
)

plt.show()
```

```python
# mpg, weight, cylinder
# cylinder의 상대적 비율을 계산한다. (정규화 : 0~1사이의 숫자로 맞추기)

cylinder_size = df.cylinders / df.cylinders.max() * 200   # 제일 큰 값으로 나누고 비율(*200)을 맞춰주기
cylinder_size
```

```python
# 연비(mpg)와 차중(weight)에 대한 산점도 그리기
df.plot(
    kind='scatter',
    x='weight',
    y='mpg',
    figsize=(10,5),
    color=df.duplicated(subset=['weight', 'mpg'], keep=False).map(
        {True: 'blue', False: 'coral'}
    ),
    alpha=0.5,       # 투명도 조절 (중복 데이터를 구분하기 위함)
    s=cylinder_size  # 점 크기
)

plt.show()
```
