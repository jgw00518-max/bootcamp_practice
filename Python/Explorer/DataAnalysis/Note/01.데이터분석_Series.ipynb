### Series
: Pandas의 기본 데이터 타입으로 데이터가 순차적으로 나열된 1차원 배열의 형태

#### Dictionary -> Series

```python
import pandas as pd
import numpy as np
```

```python
# Dictionary
dic_data = {
    'a' : 1,
    'b' : 2,
    'c' : 3
}

dic_data
```

```python
# Series로 변환
sr = pd.Series(dic_data)
sr
```

> Series는 Key (a,b,c) 와 value (1,2,3) 로 구성되어 있다.    
> DataFrame은 index와 column으로 구성되어 있다.

#### List -> Series

```python
list_data = ['2020-12-31', 3.14, 'abc', 100, True]
list_data
```

```python
sr = pd.Series(list_data)
sr
```

>Index와 Data로 구성

```python
index = sr.index
print(list(index))

value = sr.values
print(value)
```

#### Tuple -> Series
>list와 tuple의 차이점 -> tuple은 수정할 수 없다

```python
tuple_data = ('유비', '2020-12-31', '남', False)
sr = pd.Series(tuple_data, index=('이름', '생년월일', '성별', '학생여부'))
sr
```

#### Series에서 Data 선택하기

```python
# 한 개의 index 선택
print(sr.loc['이름'])
print(sr.iloc[0]) # index 번호 선택(index 이름 바꾸기 전)
```

```python
# 두 개의 index 선택
print(sr.loc['생년월일':'성별'])    # Slicing
print(sr.loc[['생년월일', '성별']]) # Indexing
print(sr.iloc[1:3])
print(sr.iloc[[1,2]])
```

```python
# 이름 ~ 성별까지 추출
# loc와 Slicing
print(sr.loc['이름':'성별'])
# loc와 Indexing
print(sr.loc[['이름', '생년월일', '성별']])
# iloc와 Slicing
print(sr.iloc[0:2+1])
# iloc와 Indexing
print(sr.iloc[[0,1,2]])
```

#### Series의 연산

```python
student = pd.Series(
    {
        '국어' : 100,
        '영어' : 80,
        '수학' : 90
    }
)

student
```

```python
# 과목의 점수를 200으로 나누자 (iteration type -> 알아서 반복)
student / 200
```

#### Series간 연산

```python
student1 = pd.Series(
    {
        '국어' : 100,
        '영어' : 80,
        '수학' : 90
    }
)

student2 = pd.Series(
    {
        '수학' : 80,
        '국어' : 90,
        '영어' : 80
    }
)

print(student1)
print(student2)
```

```python
# 덧셈
addition = student1 + student2
addition
```

> index에 맞추어져 연산이 이루어 진다.

```python
subtraction = student1 - student2
multiplication = student1 * student2
division = student1 / student2
```

```python
# 사칙연산 결과를 데이터프레임으로 만들기
result = pd.DataFrame(
    {
        '덧셈' : addition,
        '뺄셈' : subtraction,
        '곱셈' : multiplication,
        '나눗셈' : division
    }
)
result.T    # Dictionary를 이용한 방법

result = \
pd.DataFrame(
    [addition, subtraction, multiplication, division],
    index=['덧셈', '뺄셈', '곱셈', '나눗셈']
)
result      # list를 이용한 방법
```

### NaN(Not a Number)(None, Na(Not Available))값이 있는 연산

```python
student1 = pd.Series(
    {
        '국어' : np.nan,
        '영어' : 80,
        '수학' : 90
    }
)
student1
```

```python
student2 = pd.Series(
    {
        '수학' : 80,
        '국어' : 90,
    }
)
student2
```

```python
addition = student1 + student2  # 결측치가 있을 시에 연산 불가
addition
```

> 덧셈에서 객체 사이에 공통 index가 없거나 NaN이 포함된 경우 연산결과는 NaN으로 반환

```python
# 결측치 처리 방법 : 결측치 값을 0으로 대체하여 계산
addition = student1.add(student2, fill_value=0)
print(addition)

subtraction = student1.sub(student2, fill_value=0)
print(subtraction)

multiplication = student1.mul(student2,fill_value=0)
print(multiplication)

division = student1.div(student2, fill_value=0)
print(division)
```

```python
result = \
pd.DataFrame(
    [addition, subtraction, multiplication, division],
    index=['덧셈', '뺄셈', '곱셈', '나눗셈']
)
result
```
