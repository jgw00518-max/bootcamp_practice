### Seaborn - 고급 그래프
: Matplotlib의 기능과 스타일을 확장한 파이썬 시각 도구의 고급 버전

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import koreanize_matplotlib
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

#### 회귀선이 있는 산점도
: 선형회귀분석에 의한 회귀선을 생성

```python
# 스타일 테마(darkgrid, whitegrid, dark, white, ticks)
sns.set_style('darkgrid')
```

```python
fig = plt.figure(figsize=(15,5))
ax1 = fig.add_subplot(1,2,1)
ax2 = fig.add_subplot(1,2,2)

sns.regplot(
    x='age',
    y='fare',
    data=titanic,
    scatter_kws={'color' : 'black'},
    line_kws={"color" : "red"},
    ax=ax1
)
sns.regplot(
    x='age',
    y='fare',
    data=titanic,
    # scatter_kws={'color' : 'black'},
    # line_kws={"color" : "red"},
    ax=ax2,
    fit_reg=False
)

ax1.set_title("Scatter for age with regression")
ax2.set_title("Scatter for age")

plt.show()
```

> 회귀선에 옅은 색의 범위는 오차범위

---
#### 히트맵(hearmap)

```python
# 성별 객실별 승선 인원수
table = titanic.pivot_table(
    index=['sex'],
    columns=['class'],
    aggfunc='size'
)
table
```

```python
sns.heatmap(
    table,
    annot=True,
    fmt='d',        # 숫자 변환 (ex: 1.4e+02 -> 144)
    cmap='YlGnBu',  # 색상 변경
    linewidths=.5
)
plt.show()
```

----
#### 데이터의 산점도

```python
sns.set_style('whitegrid')
```

```python
fig = plt.figure(figsize=(18,6))
ax1 = fig.add_subplot(1,2,1)
ax2 = fig.add_subplot(1,2,2)

# 데이터의 분산을 고려하지 않은 분포
sns.stripplot(
    x='class',
    y='age',
    data=titanic,
    ax=ax1
)
# 데이터의 분산을 고려한 분포
sns.swarmplot(
    x='class',
    y='age',
    data=titanic,
    ax=ax2
)

ax1.set_title('Strip Plot')
ax2.set_title('Swarm Plot')

plt.show()
```

---
#### 막대 그래프

```python
fig = plt.figure(figsize=(15,5))
ax1 = fig.add_subplot(1,3,1)
ax2 = fig.add_subplot(1,3,2)
ax3 = fig.add_subplot(1,3,3)

sns.barplot(
    x='sex',
    y='survived',
    data=titanic,
    ax=ax1
)
sns.barplot(
    x='sex',
    y='survived',
    hue='class',
    data=titanic,
    ax=ax2
)
sns.barplot(
    x='sex',
    y='survived',
    hue='class',
    data=titanic,
    dodge=False,
    ax=ax3
)

plt.show()
```

> 바 위에 있는 검은색 선은 오차막대

```python
fig = plt.figure(figsize=(15,5))
ax1 = fig.add_subplot(1,3,1)
ax2 = fig.add_subplot(1,3,2)
ax3 = fig.add_subplot(1,3,3)

sns.countplot(
    x='class',
    hue='class',
    palette='Set1',
    data=titanic,
    ax=ax1,
    dodge=False,
    legend=True
)

sns.countplot(
    x='class',
    hue='who',
    palette='Set2',
    data=titanic,
    ax=ax2
)
sns.countplot(
    x='class',
    hue='who',
    palette='Set3',
    data=titanic,
    dodge=False,
    ax=ax3
)

plt.show()
```

---
#### 박스플롯/바이올린 그래프
: 박스플롯은 주요 통계 지표를 제공하지만 분산의 정도를 알 수가 없다.

```python
fig=plt.figure(figsize=(15, 10))
ax1 = fig.add_subplot(2,2,1)
ax2 = fig.add_subplot(2,2,2)
ax3 = fig.add_subplot(2,2,3)
ax4 = fig.add_subplot(2,2,4)

sns.boxplot(
    x='alive',
    y='age',
    data=titanic,
    ax=ax1
)

sns.boxplot(
    x='alive',
    y='age',
    hue='sex',
    data=titanic,
    ax=ax2
)

sns.violinplot(
    x='alive',
    y='age',
    data=titanic,
    ax=ax3
)

sns.violinplot(
    x='alive',
    y='age',
    hue='sex',
    data=titanic,
    ax=ax4
)
```

----
#### 조인트 그래프
: 산점도를 기본으로 표시하고 각 축에 대한 히스토그램을 동시 제공

```python
sns.jointplot(
    x='fare',
    y='age',
    data=titanic
)

sns.jointplot(
    x='fare',
    y='age',
    kind='reg',
    data=titanic
)

sns.jointplot(
    x='fare',
    y='age',
    kind='hex',
    data=titanic
)

sns.jointplot(
    x='fare',
    y='age',
    kind='kde',
    data=titanic
)

plt.show()
```

---
#### 조건을 적용하여 화면을 그리드로 분할하기

```python
# 조건에 따라 그리드 나누기
g = sns.FacetGrid(
    data= titanic,
    col='who',
    row='survived'
)

# 그래프 적용
g.map_dataframe(sns.histplot, x='age')

plt.show()
```
