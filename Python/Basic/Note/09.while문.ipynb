#### while문

```python
# 점수가 80점 이상이면 우수, 아니면 노력으로 출력 #1
scores = [78, 85, 95, 65]

i = 0
while i < len(scores):
    if scores[i] >= 80:
        print("우수")
    else:
        print("노력")
    i += 1
```

```python
# 점수가 80점 이상이면 우수, 아니면 노력으로 출력 #2
scores = [78, 85, 95, 65]

while scores:
    score = scores.pop(0)
    if score >= 80:
        print("우수")
    else:
        print("노력")
```

```python
# 평균 구하기
scores = [78, 85, 95, 65]

length = len(scores)
total = 0
while scores:
    total += scores.pop()

print(total/length)
```

```python
import math

# 표준편차 구하기 - 1
scores = [78, 85, 95, 65]

# 평균 구하기
total = 0
i = 0

while i < len(scores):
    total += scores[i]
    i += 1

avg = total / len(scores)
print(avg)

# 각 데이터와 평균의 차이의 제곱의 합의 평균 구하기(분산)
total = 0
i = 0

while i < len(scores):
    total += (scores[i] - avg) ** 2
    i += 1

avg = total / len(scores)
print(avg)

# 제곱근 씌우기(표준편차)
math.sqrt(avg)
```

#### while else
: else문을 반복문이 정상적으로 종료 되었을 경우에 실행이 된다.    
while문 안에 break가 있어 반복이 중간에 멈추었을 경우에는 else문은 실행되지 않는다.

```python
i = 1
while i < 5:
    print(i)
    i += 1
else:
    print("반복 종료")
```

```python
i = 1
while i < 5:
    print(i)
    if i == 2:
        break
    i += 1
else:
    print("반복 종료")
```
