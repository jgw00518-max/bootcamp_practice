#### 내장함수

```python
# abs : 절댓값
print(abs(3))
print(abs(-3))
```

```python
# all : 자료가 모두 참일 경우에만 True
print(all([1,2,-3]))
print(all([1,2,0]))
```

```python
# any : 자료 중 하나라도 참일 경우에는 True
print(any([0,0,1]))
print(any([0,0,0]))
print(any([0,"",""]))
print(any([]))
```

```python
# ord : 문자 -> ASCII Code로 변환
print(ord('A'))
print(ord('Z'))
print(ord('a'))
print(ord('가'))
print(ord('!'))
print(ord('1'))
```

```python
# chr : ASCII Code값 -> 문자로 출력
print(chr(65))
print(chr(90))
print(chr(97))
print(chr(44032))
print(chr(33))
print(chr(49))
```

```python
# divmod : 나눈 값의 결과를 몫과 나머지로 구분하여 출력
divmod(7,3)
```

```python
# enumerate : index를 표시해준다.
for i, name in enumerate(['aaa', 'bbb', 'ccc']):
    print(i+1, name)
```

```python
# zip : 동일한 갯수로 이루어진 자료형을 묶어주는 역할
a = [1,2,3]
b = [4,5,6]
for i, j in zip(a, b):
    print(i ,j)
```
