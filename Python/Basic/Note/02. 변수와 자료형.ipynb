### 변수
계산에 필요한 자료를 일시적으로 저장하거나 처리결과를 담을 수 있는 기억장소를 지정해 주는 역할.    
변수 이름은 값을 지정하는 메모리 영역의 이름으로 할당.    

#### 변수 이름의 작성 규칙
- 첫 자는 영문으로 시작
- 첫 자를 제외하고 숫자, under bar, dot을 사용할 수 있음
- 대문자와 소문자를 구분
- 변수 이름은 의미를 파악할 수 있는 이름으로 지정하는 것이 좋다.
- 두 단어를 포함한 Camel 기법, Snake 기법으로 하는 것이 가독성이 좋다.

#### 문자열 관련 함수

```python
str1 = "I have a dream."

# 전체 알파벳 갯수
print(len(str1))

# a가 몇개?
print(str1.count('a'))

# have가 몇개
print(str1.count('have'))

# 문자 위치 알려주기
print(str1.find('d'))
print(str1[9:])
print(str1.find('x'))
print(str1.find('a'))
```

```python
# 특정 글자의 모든 위치를 알아내는 방법
str1 = "I have a dream."
search_word = 'a'
start = 0
count = 0

while True:
    position = str1.find(search_word, start)
    if position == -1:
        break
    count+=1
    print(f"단어 {search_word}의 {count}번의 위치는 {position} 입니다.")
    start = position + len(search_word)
```

```python
# 정규식을 이용한 특정 글자의 모든 위치를 알아내는 방법 (느린 방법)
import re

str1 = "I have a dream."
search_word = 'a'

position = []

for match in re.finditer(search_word, str1):
    position.append(match.start())

print(f"단어 {search_word}의 위치는 {position} 입니다.")
```

```python
# 정규식을 이용한 특정 글자의 모든 위치를 알아내는 방법 (List Comprehension)
import re

str1 = "I have a dream."
search_word = 'a'

position = [match.start() for match in re.finditer(search_word, str1)]

print(f"단어 {search_word}의 위치는 {position} 입니다.")
```
