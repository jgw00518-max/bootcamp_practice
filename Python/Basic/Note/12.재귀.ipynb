#### 재귀(Recusion)
- 계산을 위해 자기 자신을 참조하는 방법
- Factorial : 3! = 3 * 2 * 1

#### 반복 방식의 Factorial 함수

```python
def fact_iter(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result
```

```python
print(fact_iter(3))
```

#### 재귀방식의 Factorial 함수

```python
def fact_rec(n):
    if n == 1:
        return n
    else:
        return n * fact_rec(n-1)
```

```python
print(fact_rec(3))
```
