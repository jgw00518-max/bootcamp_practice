#### 딕셔너리(Dictionary)
: Dictionary는 List나 Tuple처럼 순차적으로 해당 index를 구하지 않고 Key를 통해 Value를 얻는다.

```python
# 딕셔너리 생성
dic = {
    'name' : 'petty',
    'phone' : '010-1111-2222',
    'birth' : '1231'
}
dic
```

```python
dic['name']
```

```python
# Dictionary에 Value가 List인 경우
dic = {
    '성명' : ['유비', '관우', '장비'],
    '전화번호' : [111, 222, 333]
}
dic
```

```python
# 유비만 추출하기
dic['성명'][0]
```

```python
# 데이터 추가하기
dic['주소'] = '서울'
dic
```

```python
# 데이터 수정하기
dic['주소'] = '인천'
dic
```

```python
# 삭제하기
del dic['주소']
dic
```

```python
# Key 값 확인
list(dic.keys())
```

```python
# temp에서 유비 -> 111로 출력하기
temp = list(dic.values())
print(f"{temp[0][0]} -> {temp[-1][0]}")
```

### Ex

```python
d = {
    'name' : '홍길동',
    'age' : 30
}
d
```

```python
# 내 이름은 홍길동이고 나이는 30세 입니다.
print(f"내 이름은 {d['name']}이고 나이는 {d['age']}세 입니다.")
```
