#### Class

```python
# 1) Property를 통한 Data 전달

class CalcProperty():
    # Property
    firstNum = 0
    secondNum = 0

    # 생성자
    # Method
    def add(self):
        return self.firstNum + self.secondNum
```

```python
# Class의 사용
calcProperty = CalcProperty()
calcProperty.firstNum = 10
calcProperty.secondNum = 20
calcProperty.add()
```

```python
# 2) 함수를 통한 Data 전달
class CalcFunction():
    # Property
    # constructor
    # method
    def add(self, first, second):
        return first + second
```

```python
calcFunction = CalcFunction()
calcFunction.add(10, 20)
```

```python
# 3) 생성자를 통한 Data 전달

class CalcConst():
    # Property
    firstNum = 0
    secondNum = 0

    # constructor
    def __init__(self, firstNum, secondNum):
        self.firstNum = firstNum
        self.secondNum = secondNum

    # method
    def add(self):
        return self.firstNum + self.secondNum
```

```python
calcConst = CalcConst(10, 20)
calcConst.add()
```

```python
class Calc():
    # Property
    firstNum = 0
    secondNum = 0

    # constructor
    def setdata(self,firstNum,secondNum):
        self.firstNum = firstNum
        self.secondNum = secondNum

    # method
    def sum(self):
        return self.firstNum + self.secondNum
    def mul(self):
        return self.firstNum * self.secondNum
    def sub(self):
        return self.firstNum - self.secondNum
    def div(self):
        return self.firstNum / self.secondNum
        
```

```python
calc01 = Calc()
calc01.setdata(4,2)
print('덧셈:', calc01.sum())
print('곱셈:', calc01.mul())
print('뺄셈:', calc01.sub())
print('나눗셈:', calc01.div())
```

Class 상속

```python
class Add():
    num1 = 0
    num2 = 0

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        result = self.num1 + self.num2
        return result
```

```python
addition = Add(1,2)
print(addition.add())
```

```python
class SubAdd(Add):
    num1 = 0
    num2 = 0

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def sub(self):
        result = self.num1 - self.num2
        return result
```

```python
subAdd = SubAdd(2,3)
print(subAdd.add())
print(subAdd.sub())
```

#### 다중 상속

```python
# 첫 번째 상위 클래스
class Flyable:
    def fly(self):
        print("I'm flying.")
```

```python
# 두 번째 상위 클래스
class Swimmable:
    def swim(self):
        print("I'm swimming.")
```

```python
# 하위 클래스
class Duck(Flyable, Swimmable):
    def quack(self):
        print("Quack quack!")
```

```python
my_duck = Duck()
my_duck.fly()
my_duck.swim()
my_duck.quack()
```
