### 리스트(list) 자료형

#

```python
# 리스트 자료형 선언
var01 = []
var02 = [1,2,3]
var03 = ['I', 'have', 'a', 'dream']
var04 = [1, 2, 'Dream', 'True']
var05 = [1, 2, ['Dream', 'True']]
```

---
#### 리스트의 연산자

```python
# 리스트의 숫자 합치기
var01 = [1, 2, 3]
var02 = [4, 5, 6]

var01 + var02
```

```python
print(var01[0] + var02[0])
print(var01[1] + var02[1])
print(var01[2] + var02[2])
```

```python
var03 = []
for i in range(0, len(var01)):
    var03.append(var01[i] + var02[i])

var03
```

```python
# List Comprehension
var03 = [var01[i] + var02[i] for i in range(0, len(var01))]
var03
```

```python
# 리스트의 숫자 반복
var01 * 3
```

```python
# 문자열 리스트 합치기
var01 = ['a', 'b', 'c']
var02 = ['d', 'e', 'f']
var01 + var02
```

```python
# Data 존재 여부 확인
list01 = [1, 2, 3 ,4 ,5]
print(3 in list01)
print(6 in list01)
```

### Ex


```python
# evenVar에 2부터 100까지의 짝수 List를 생성
evenVar = list(range(2, 100+1, 2))
# print(evenVar)

# oddVar에 1부터 99까지의 홀수 List를 생성
oddVar = list(range(1, 99+1, 2))
# print(oddVar)

# evenVar에 oddVar 확장하기
evenVar.extend(oddVar)
# print(evenVar)

# evenVar를 정렬하기
evenVar.sort()
print(evenVar)
```

```python
# 갯수 구하기
len(evenVar)

# 합계 구하기
sum(evenVar)
```

```python
evenVar.append('a')
# sum(evenVar)
type(evenVar)
```

### Ex

```python
# f-string을 이용하여 Hello World! My name is Alice and I'm 30 years old. 를 추출하기
list01 = [1, 3.14, 'Hello World!', True, [170, 50, 30], ['Alice', 'Cathy']]

print(f"{list01[2]} My name is {list01[-1][0]} and I'm {list01[-2][-1]} years old.")
```

#### 리스트의 복사

```python
# Call by reference
var01 = [1,2,3,4]
var02 = var01

var02[0] = 10
print(var02)
print(var01)
```

```python
# Call by value
var01 = [1,2,3,4]
var02 = var01.copy()

var02[0] = 10
print(var02)
print(var01)
```
