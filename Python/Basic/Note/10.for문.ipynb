### for문
: for 변수 in Data(list, tuple, 문자열, range())

```python
# 1 ~ 10 출력하기
for i in range(1, 10+1):
    print(i)
```

```python
# Ex : 1 ~ 10까지의 수 중 홀수만 출력하기
for i in range(1, 10+1, 2):
    print(i)
```

```python
# 1 ~ 10까지의 제곱근을 출력
import math
for i in range(1, 10+1):
    print(f"{i:0>2} -> {math.sqrt(i)}")
```

```python
# 1 ~ 100까지의 합
start = 1
stop = 100

total = 0
for i in range(start,stop+1):
    total += i
print(f"{start}부터 {stop}까지의 합은 {total}입니다.")
```

```python
# Ex : 1부터 1000까지의 홀수의 합과 짝수 합을 구하기
# 출력 : 짝수의 합은 250500이고 홀수의 합은 250000입니다.

evenSum = 0
oddSum = 0

for i in range(1,1000+1):
    if i % 2 == 0:
        evenSum += i
    else:
        oddSum += i
print(f"짝수의 합은 {evenSum:,}이고 홀수의 합은 {oddSum:,}입니다.")
```

```python
print(sum(range(1, 1000+1, 2)))
print(sum(range(2, 1000+1, 2)))
```

```python
# 구구단 7단
dan = 7

for i in range(1, 9+1):
    print(f"{dan} X {i} = {dan*i}")
```

```python
# 구구단 전체 출력하기 (2단 ~ 9단)

for dan in range(2, 9+1):
    print(f"===={dan}단====")
    for i in range (1, 9+1):
        print(f"{dan} X {i} = {dan*i:>2}")
```

```python
# 구구단 중 짝수단과 짝수 곱해지는 수만 출력 (단, if문 사용 불가)
for dan in range(2, 9+1, 2):
    print(f"===={dan}단====")
    for i in range (2, 9+1, 2):
        print(f"{dan} X {i} = {dan*i:>2}")
```

총 5명의 학생이 시험을 보았는데 시험 점수가 60점을 넘으면 합격이고    
그렇지 않으면 불합격이다.    


```python
# 합격, 불합격 보여주기
scores = [90, 25, 67, 45, 80]
student = 0

for score in scores:
    result = "이므로 합격" if score >= 60 else "이어서 불합격"
    student += 1
    print(f"{student}의 학생은 점수가 {score}{result}입니다.")
```

```python
# 학생들 점수의 평균을 구하여 평균 점수보다 낮으면 불합격 아니면 합격을 위의 출력과 동일하게 하기
import numpy as np
scores = [90, 25, 67, 45, 80]
student = 0

print(f"평균 점수는 {np.mean(scores)}")

for score in scores:
    result = "이므로 합격" if score >= np.mean(scores) else "이어서 불합격"
    student += 1
    print(f"{student}의 학생은 점수가 {score}{result}입니다.")
```

```python
# 60점이 넘으면 합격이다. 합격인 학생 번호만 출력

# 합격, 불합격 보여주기
scores = [90, 25, 67, 45, 80]
student = 0

for score in scores:
    if score >= 60:
        result = "합격"
        student += 1
        print(f"{student}의 학생은 {result}입니다.")
    else:
        student += 1
```

```python
# 60점 이상인 합격자 명단 List 제작
scores = [90, 25, 67, 45, 80]
list01 = [i + 1 for i in range(len(scores)) if scores[i] >= 60]
print(list01)
```

#### 문장과 단어 추출


```python
str01 = """나는 홍길동 입니다.
주소는 서울시 입니다.
나이는 35세 입니다."""
```

```python
str01
```

```python
sents = [] # 문장 저장
words = []
# 문단 -> 문장
for sen in str01.split("\n"):
    sents.append(sen)
    for word in sen.split():
        words.append(word)

print(sents)
print(words)
print(f"문장 수 : {len(sents)}")
print(f"단어 수 : {len(words)}")
```

```python
# 중복 되지 않는 단어의 수 구하기
len(list(set(words)))
```

#### for else
- for문이 모두 실행되면 else문이 실행된다.
- for문이 break를 만나면 else문은 실행되지 않는다.

```python
fruits = ['apple', 'banana', 'cherry', 'orange']
search = 'orange'

for fruit in fruits:
    if fruit == search:
        print(f"{search}를 찾았습니다.")
        break
else:
    print(f"{search}가 없습니다.")
```

#### 파이썬의 내포(Comprehension)

```python
# 기존 방법
var01 = [1,2,3,4]
var02 = []
for i in var01:
    var02.append(i*5)
var02
```

```python
# List Comprehension
var01 = [1,2,3,4]
var02 = [i*5 for i in var01]
var02
```

```python
# 기존 방법
var01 = [1,2,3,4]
var02 = []
for i in var01:
    if i%2 == 0:
        var02.append(i*5)
var02
```

```python
# List Comprehension
var02 = [i*5 for i in var01 if i%2==0]
var02
```

```python
# Ex : 구구단 결과만 리스트로 만들기 (2~9단, 단 : List Comprehension 사용)

guguDan = [dan*i for dan in range (2, 9+1) for i in range(1, 9+1)]
print(guguDan)
```

```python
# Ex : 60점 이상의 학생 점수만 List로 만들어 평균 구하기
# 출력 : 합격한 학생들의 평균 점수는 ___점 입니다.
import numpy as np
scores = [90, 25, 67, 45, 80]

print(f"합격한 학생들의 평균 점수는 {np.mean([score for score in scores if score >= 60])}점 입니다.")
```
