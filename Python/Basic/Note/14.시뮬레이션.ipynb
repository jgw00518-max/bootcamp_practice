#### 동전을 던졌을 때, 앞면과 뒷면이 나올 확률이 동일하다는 것을 증명

```python
# 난수
import random
```

```python
random.randint(1, 10) # 1 ~ 10 사이의 난수 1개 발생
```

```python
# 난수를 통한 방법
x = random.randint(1, 2)
if x == 1:
    result = '앞면'
else:
    result = '뒷면'


print(result)
```

```python
# 동전의 앞면과 뒷면 횟수 결과를 함수로 만들기
def coinF(n):
    coinFront = coinRear = 0
    for i in range(1, n+1):
        x = random.randint(1,2)
        if x == 1:
            coinFront += 1
        else:
            coinRear += 1
    print(f"{n:>7}번 시도 시 동전의 앞면은 {coinFront:>6}번이고 뒷면은 {coinRear:>6}번 입니다.")
```

```python
coinF(10)
coinF(100)
coinF(1000)
coinF(10000)
coinF(100000)
coinF(1000000)
```

#### 주사위 번호가 나올 확률이 같다는 것을 증명

(단, if문 사용금지)

```python
def dice(n):
    result = [0,0,0,0,0,0]
    for _ in range(1, n+1):
        x = random.randint(1,6)
        result[x-1] += 1
        
    print(f"{n:>7}번 굴릴 시 ", end=', ')
    for i in range(6):
        print(f"{result[i]:>6}", end=', ')
    print()
```

```python
dice(10)
dice(100)
dice(1000)
dice(10000)
dice(100000)
dice(1000000)
```
