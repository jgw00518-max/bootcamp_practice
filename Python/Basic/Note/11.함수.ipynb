#### 가변길이 인수를 가진 함수 (UnPacking 연산자)

```python
def sumManyF(*args):
    sum = 0
    for i in args:
        sum+=i
    return sum
```

```python
sumManyF(1,2,3)
```

```python
sumManyF(1,2,3,4,5,6,7)
```

#### 덧셈기능과 곱셈기능을 구분하여 숫자의 결과를 보여주는 함수

```python
def sum_mulF(choice, *inputValue):
    result = 0
    if choice == "add":
        result = np.sum(inputValue)
    elif choice == "mul":
        result = np.prod(inputValue)
    else:
        result = "what are you talking about?"
    return result
```

```python
sum_mulF('add', 1, 2, 3, 4, 5, 6)
```

```python
sum_mulF('mul', 1, 2, 3, 4, 5, 6)
```

```python
# what are you talking about?
sum_mulF('sum', 1, 2, 3, 4, 5, 6)
```

#### BMI 계산

```python
def bmi(height, weight):
    index = weight / (height / 100)**2
    if index < 18.5:
        result = "저체중"
    elif index < 23:
        result = "정상"
    elif index < 25:
        result = "과체중"
    elif index < 30:
        result = "비만"
    else:
        result = "고도비만"
    return index, result
```

```python
# 귀하의 bmi 지수는 15.6이며 저체중 입니다.
index, result = bmi(170,59)
print(f"귀하의 bmi 지수는 {round(index,1)}이며 {result}입니다.")
```

```python
# Ex : 숫자를 입력 받아 해당 숫자까지 제곱값을 출력하는 함수를 작성
def print_squares(x):
    # for i in range(x+1):
    #     print(f"{i} -> {i**2}")
    return [f"{i} -> {i**2}" for i in range(x+1)]
```

```python
print_squares(3)
```

```python
import math

def circle_area(radius):
    return math.pi * radius ** 2

circle_area(5)
```

```python
# Ex : 원의 면적과 테두리 길이를 구하는 함수를 작성
def circle_area_length(x):
    area = math.pi * x ** 2
    length = 2 * math.pi * x
    return area, length
```

```python
area, length = circle_area_length(5)
print("원의 면적 : ", area, ", 원의 길이 : ", length)
```

#### 중첩 함수(nested function) / 내부 함수(inner function)
: 함수 안에서 다른 함수를 정의

```python
def outer_function(x, y):
    def inner_function(z):
        return x + y + z
    return inner_function
```

```python
result = outer_function(1,2)
print(result(3))
print(result(4))
```

```python
result = outer_function(3,4)
print(result(3))
print(result(4))
```
