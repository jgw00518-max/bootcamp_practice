#### map 함수
: 여러개의 데이터(리스트 등)에 함수를 한번에 적용해 주는 함수    
- 반복문(for)으로 하나씩 처리하던 작업을 한 줄로 간단하게 표현
- 형태 : map(함수, 반복가능한 자료형)
- list(map(function, iterable data))로 많이 사용

---
### for 와 map의 비교

```python
# for문 : 모든 수자에 2를 곱하기
numbers = [1, 2, 3, 4]
result = []

for n in numbers:
    result.append(n*2)

print(result)
```

```python
# map : 일반 함수를 이용한 모든 숫자에 2를 곱하기
numbers = [1, 2, 3, 4]

def numSquare(n):
    return n*2

result = list(map(numSquare, numbers))
print(result)
```

```python
# map : lambda 함수를 이용한 모든 숫자에 2를 곱하기
numbers = [1,2,3,4]
result = list(map(lambda x: x*2,numbers))
print(result)
```

### Ex

```python
# for : 문자열을 대문자로 변환
words = ['apple', 'banana', 'cherry']
result = []

# for w in words:
#     result.append(w.upper())

[result.append(w.upper()) for w in words]

print(result)
```

```python
# map : 일반함수를 이용한 문자열을 대문자로 변환
words = ['apple', 'banana', 'cherry']
def upperWord(w):
    return w.upper()

result = list(map(upperWord, words))
print(result)
```

```python
# map : lambda 함수를 이용한 문자열을 대문자로 변환
words = ['apple', 'banana', 'cherry']
result = list(map(lambda w: w.upper(), words))
print(result)
```

```python
# map : 제공하는 함수를 이용한 문자열을 대문자로 변환
words = ['apple', 'banana', 'cherry']
result = list(map(str.upper,words))
print(result)
```

#### 숫자를 문자열로 변환

```python
# for문
nums = [1,2,3]
result = []

# for n in nums:
#     result.append(str(n))

[result.append(str(n)) for n in nums]
print(result)
```

```python
# map : 일반 함수 사용
nums = [1,2,3]
def numStr(n):
    return str(n)
result = list(map(numStr, nums))
print(result)
```

```python
# map : 제공하는 함수 사용
nums = [1,2,3]
result = list(map(str, nums))
print(result)
```

```python
# lambda 사용
nums = [1,2,3]
result = list(map(lambda n: str(n), nums))
print(result)
```

#### 리스트의 Data 길이 구하기

```python
# for
words = ['hi', 'hello', 'python']
result = []

# for n in words:
#     result.append(len(n))

[result.append(len(n)) for n in words]
print(result)
```

```python
# map : 일반 함수 제작
words = ['hi', 'hello', 'python']
def numLen(n):
    return len(n)

result = list(map(numLen, words))
print(result)
```

```python
# 제공되는 함수
words = ['hi', 'hello', 'python']
result = list(map(len,words))
print(result)
```

```python
# lambda 함수를 이용한 이름 뒤에 '님!' 붙이기
names = ['철수', '영희', '민수']
result = list(map(lambda x: x + '님!', names))
print(result)
```

```python
# 제공되는 함수를 이용한 문자열을 숫자로 변환
nums= ['10', '20', '30']
result = list(map(int,nums))
print(result)
```

```python
# map과 lambda함수를 이용한 2개의 리스트의 요소별 합계 구하기
a = [1,2,3]
b = [10,20,30]
result = list(map(lambda x, y : x + y, a, b))
print(result)
```
