#### 자료형의 참과 거짓
- 문자열인 경우 ""안에 data가 없으면 거짓, 있으면 참
- 리스트인 경우 [] ->
- 튜플인 경우 () ->
- 딕셔너리인 경우 {} ->
- 숫자인 경우 0이면 거짓, 0이 아니면 참

```python
var01 = [1,2,3,4]
while(var01):
    print(var01.pop(0))

var01
```

```python
var01 = 1
var02 = 2

var01, var02 = 1, 2
var01 = var02 = 1
print(var01, var02)

```

```python
(var01, var02) = (1, 'Lee')
print(var01, var02)
```

```python
# 변수값 교환
var01 = 3
var02 = 5

# temp = var01
# var01 = var02
# var02 = temp

# print(var01, var02)

var01, var02 = var02, var01
print(var01, var02)
```

#### 리스트의 복사

```python
var01 = [1,2,3]
var02 = var01.copy()
var02[1] = 4
print(var01, var02)
```

```python
var01 = [1,2,3]
var02 = var01[:]
var02[1] = 4
print(var01, var02)
```

```python
from copy import copy
var01 = [1,2,3]
var02 = copy(var01)
var02[1] = 4
print(var01, var02)
```
