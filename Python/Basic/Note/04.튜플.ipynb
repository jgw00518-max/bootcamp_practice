### 튜플(Tuple) 자료형
: 튜플은 다음과 같은 점을 제외하면 리스트와 거의 동일하다.
- ( )로 사용
- 리스트는 그 값을 생성, 삭제, 수정이 가능하지만 튜플은 값을 변경할 수 없다.    
프로그램 작성 시 생성되는 값을 변화하면서 사용하고자 하면 리스트를 사용하고    
한번 생성된 데이터 값이 변경되는 것을 막고자 하면 튜플을 사용.    

```python
var01 = ()
var02 = (1,) # 데이터가 하나여도 반드시 ,를 붙인다.
var03 = (1,2,3)
var04 = 1,2,3 # 괄호가 없어도 tuple
var05 = ('a', 'b', ('ab', 'cd'))

type(var04)
```

```python
var05[-1][0]
```

#### Tuple 값 변경
: Tuple에 List를 추가하여 사용

```python
tuple01 = ([1, 2], 3, 4)
tuple01[0][0] = 100 # tuple 안에 있는 list는 변경 가능
tuple01

list01 = list(tuple01) # tuple 안에 있는 값을 바꾸고 싶을 땐, list로 변경 후 값을 변경
list01[1] = 30
tuple(list01)
```
