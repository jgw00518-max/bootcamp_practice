### if 문

```python
x = 10

if x > 50:
    print(f"{x}는 50보다 큰 수 입니다.")
else:
    print(f"{x}는 50보다 같거나 작은 수 입니다.")

print(">>> End <<<")
```

```python
x = 100

if x > 50:
    print(f"{x}는 50보다 큰 수 입니다.")
elif x < 50:
    print(f"{x}는 50보다 작은 수 입니다.")
else:
    print(f"{x}는 50과 같은 수 입니다.")

print(">>> End <<<")
```

```python
# 사용자로부터 입력 받기
int(input("숫자를 입력 하세요 : "))
```

```python
# 사용자로부터 숫자를 입력 받아 짝수 홀수 구분하기
if int(input("숫자를 입력 하세요 : ")) % 2 == 0:
    print("짝수")
else:
    print("홀수")
```

```python
# 사용자로부터 시험 점수를 입력 받아 학점으로 표현하기
# (입력 시험점수의 범위를 꼭 확인하기, 단: print()는 2개만 사용)
score = int(input("시험 점수를 입력하세요 : "))

if 0 <= score <= 100:
    if score >= 90:
        grade = 'A'
    elif score >= 80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60:
        grade = 'D'
    else:
        grade = 'F'
    print(f"학점은 : {grade} 입니다.")
else:
    print("점수는 0~100점까지만 입력가능합니다.")

```

#### match case
: 3.10부터 도입

```python
x = int(input('숫자를 입력 하세요'))

match x % 2:
    case 0:
        print(f"{x}는 짝수 입니다.")
    case 1:
        print(f"{x}는 홀수 입니다.")
```

```python
x = int(input('숫자를 입력 하세요'))

match x:
    case x if x % 2 == 0:
        print(f"{x}는 짝수 입니다.")
    case _:
        print(f"{x}는 홀수 입니다.")
```

```python
var01 = int(input("숫자를 입력 하세요."))

match var01:
    case 0:
        print("0")
    case 88 | 99:
        print("88이거나 99")
    case var01 if var01 % 10 == 0:
        print("10의 배수")
    case var01 if var01 > 0:
        print("양수")
    case var01 if var01 < 0:
        print("음수")
    case _:
        print("~~~~")
```

```python
# 학점을 match case로 변환

score = int(input("학점을 입력 하세요."))

if 0 <= score <= 100:
    match score // 10:
        case 10 | 9:
            grade = 'A'
        case 8:
            grade = 'B'
        case 7:
            grade = 'C'
        case 6:
            grade = 'D'
        case _:
            grade = 'F'
    print(f"학점은 {grade}입니다.")
else:
    print("점수는 0~100 사이로 입력하세요.")
```

```python
# List의 내용을 if문으로 확인하기

pocket = ['paper', 'smartphone']
card = 0

if 'money' in pocket:
    print("현금으로 택시타기")
elif card:
    print("카드로 택시타기")
else:
    print("다리로 걷기")
```

#### 삼항연산자

```python
# 짝수 홀수 판단
num = int(input("숫자를 입력하세요 : "))

result = "짝수" if num % 2 == 0 else "홀수"
result
```

```python
# Ex : 사용자로부터 정수를 입력 받아 5보다 큰 수이면 입력 받은 수 *2를 하고 아니면 입력 받은 수 +2를 한다.
num = int(input("숫자를 입력하세요 : "))

result = num * 2 if num >= 5 else num + 2
result
```
