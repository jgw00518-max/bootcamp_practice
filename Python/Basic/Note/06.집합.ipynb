### 집합(Set) 자료형
- 집합 자료형은 집합에 관련된 것들을 쉽게 처리한다.
- 중복을 허용하지 않는다.
- 순서가 없다.

```python
# set의 생성
s0 = {1,2,3,4}
s0
```

```python
# 집합 구성하기
temp = [1,2,3,4,5,6]
s1 = set(temp)
s2 = set([4,5,6,7,8,9])
s2
```

```python
# 교집합
s1 & s2
```

```python
# 합집합
s1 | s2
```

```python
# 차집합
s1 - s2
```

```python
# List의 중복값 제거 하기
var01 = [1,1,2,2,3,3,4,4]
var02 = list(set(var01))
var02
```
